### Conceptos iniciales
Una **interfaz** reúne los elementos mediante los cuales una persona utiliza el programa. Un **contenedor** agrupa controles. Un **gestor de geometría** decide dónde se ubican. Un **estilo** define aspectos visuales de los widgets `ttk`. La **validación** comprueba si los datos cumplen las reglas antes de utilizarlos.

**Distingue:** un campo bien alineado puede contener un dato incorrecto. El diseño y la validación resuelven necesidades diferentes.

## 2. Contenedores y distribución 
### 2.1. `Frame` y `LabelFrame`
`ttk.Frame` agrupa controles sin título visible. `ttk.LabelFrame` agrupa controles dentro de un borde con un título. Ambos permiten dividir una ventana en secciones.

El primer argumento de un widget es su **contenedor padre**. `ttk.Entry(cliente)` pertenece al contenedor `cliente`; `ttk.Entry(ventana)` pertenece directamente a la ventana. Esta elección determina dentro de qué espacio se organiza.

### 2.2. La cuadrícula de `grid()`
`grid()` organiza los controles en filas y columnas. Los índices comienzan en cero.

| Parámetro | Función | Ejemplo |
|---|---|---|
| `row` | Fila | `row=0` |
| `column` | Columna | `column=1` |
| `columnspan` | Cantidad de columnas ocupadas | `columnspan=2` |
| `padx`, `pady` | Espacio exterior horizontal y vertical | `padx=8, pady=5` |
| `sticky` | Alineación o expansión dentro de la celda | `sticky="w"` |

`n`, `s`, `e` y `w` representan norte, sur, este y oeste. `w` alinea a la izquierda; `e` a la derecha; `ew` estira horizontalmente; `nsew` estira en ambas direcciones.

**Regla:** no uses `pack()` y `grid()` para organizar widgets que comparten el mismo padre. Sí puedes usar `pack()` para colocar un contenedor en la ventana y `grid()` para distribuir los controles dentro de ese contenedor. En nuestros ejemplos usaremos `grid()`.

### Ejemplo 1: un formulario dividido en secciones
Antes de ejecutarlo, identifica los padres de las etiquetas y entradas.

```python
import tkinter as tk
from tkinter import ttk

ventana = tk.Tk()
ventana.title("Pedido: formulario organizado")

principal = ttk.Frame(ventana, padding=15)
principal.grid(row=0, column=0)

ttk.Label(principal, text="Nuevo pedido", font=("Arial", 16, "bold")).grid(
    row=0, column=0, sticky="w", pady=(0, 12))

cliente = ttk.LabelFrame(principal, text="Datos del cliente", padding=10)
cliente.grid(row=1, column=0, sticky="ew", pady=5)
ttk.Label(cliente, text="Nombre:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
ttk.Entry(cliente, width=30).grid(row=0, column=1, padx=5, pady=5)
ttk.Label(cliente, text="Teléfono:").grid(row=1, column=0, sticky="w", padx=5, pady=5)
ttk.Entry(cliente, width=30).grid(row=1, column=1, padx=5, pady=5)

producto = ttk.LabelFrame(principal, text="Datos del producto", padding=10)
producto.grid(row=2, column=0, sticky="ew", pady=5)
ttk.Label(producto, text="Producto:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
ttk.Entry(producto, width=30).grid(row=0, column=1, padx=5, pady=5)
ttk.Label(producto, text="Cantidad:").grid(row=1, column=0, sticky="w", padx=5, pady=5)
ttk.Entry(producto, width=30).grid(row=1, column=1, padx=5, pady=5)

ttk.Label(principal, text="Todavía no calculamos el pedido.").grid(row=3, column=0, pady=10)
ventana.mainloop()
```

### Lectura paso a paso
1. `Tk()` crea la ventana principal.
2. `principal` proporciona un margen interior de 15 unidades mediante `padding`.
3. `cliente` y `producto` pertenecen a `principal` y ocupan filas diferentes.
4. Dentro de cada sección, la columna 0 contiene etiquetas y la 1 contiene entradas.
5. `padx` y `pady` separan controles; `padding` separa el contenido del borde de un contenedor.

## 3. Diseño adaptable 
Una ventana adaptable redistribuye el espacio cuando cambia de tamaño. `geometry()` define el tamaño inicial; `minsize()` establece el mínimo.

### 3.1. Dos condiciones para crecer
1. La fila o columna del contenedor debe recibir espacio extra con `weight`.
2. El widget debe ocupar ese espacio usando `sticky`.

`columnconfigure(1, weight=1)` permite que la columna 1 reciba espacio horizontal adicional. Una columna con `weight=0` conserva el tamaño que necesita. Si dos columnas tienen pesos 1 y 2, el espacio adicional se reparte en esa proporción.

Hay que configurar **cada nivel de contenedores**: ventana → contenedor principal → sección. Si un nivel queda fijo, puede impedir que el formulario se amplíe.

### Ejemplo 2: entradas que crecen y resumen expansible

```python
import tkinter as tk
from tkinter import ttk

ventana = tk.Tk()
ventana.title("Formulario adaptable")
ventana.geometry("620x360")
ventana.minsize(440, 280)
ventana.columnconfigure(0, weight=1)
ventana.rowconfigure(0, weight=1)

principal = ttk.Frame(ventana, padding=15)
principal.grid(row=0, column=0, sticky="nsew")
principal.columnconfigure(0, weight=1)
principal.rowconfigure(1, weight=1)

formulario = ttk.LabelFrame(principal, text="Cliente", padding=10)
formulario.grid(row=0, column=0, sticky="ew")
formulario.columnconfigure(1, weight=1)
for fila, titulo in enumerate(("Nombre:", "Teléfono:")):
    ttk.Label(formulario, text=titulo).grid(row=fila, column=0, sticky="w", padx=5, pady=5)
    ttk.Entry(formulario).grid(row=fila, column=1, sticky="ew", padx=5, pady=5)

resumen = ttk.LabelFrame(principal, text="Observaciones", padding=10)
resumen.grid(row=1, column=0, sticky="nsew", pady=(10, 0))
resumen.columnconfigure(0, weight=1)
resumen.rowconfigure(0, weight=1)
tk.Text(resumen, height=5, width=30, wrap="word").grid(row=0, column=0, sticky="nsew")
ventana.mainloop()
```

### Experimenta y explica
- Amplía la ventana: las entradas crecen horizontalmente y el cuadro de observaciones en ambas direcciones.
- Cambia `sticky="ew"` de una entrada por `sticky="w"`: la columna puede crecer, pero la entrada conserva su ancho.
- Quita temporalmente `formulario.columnconfigure(1, weight=1)`: observa cómo se distribuye el espacio.
- Restablece el ejemplo y reduce la ventana hasta su tamaño mínimo.

**Idea central:** `weight` distribuye espacio en el contenedor; `sticky` determina cómo lo utiliza el widget.

**Error frecuente:** configurar las columnas de la ventana cuando las entradas pertenecen a un `LabelFrame`. Debes configurar las columnas del padre directo de esas entradas.

## 4. Apariencia con `ttk.Style` 
Un estilo permite reutilizar una apariencia. En lugar de definir el formato de cada título por separado, creamos un estilo y lo aplicamos a varias etiquetas.

| Operación | Uso |
|---|---|
| `ttk.Style(ventana)` | Crear el objeto que administra estilos |
| `.configure("Titulo.TLabel", ...)` | Definir propiedades del estilo |
| `style="Titulo.TLabel"` | Aplicarlo a una etiqueta |
| `.map(...)` | Definir propiedades según el estado del widget |

Los nombres personalizados terminan con la clase del widget: `TLabel`, `TButton` o `TEntry`. Usa estilos propios para no modificar accidentalmente todos los controles.

Los temas del sistema pueden interpretar colores y bordes de manera diferente. La legibilidad también depende de títulos claros, separación y tamaños. En este ejemplo usamos fuentes y márgenes, que son suficientes para mejorar la organización.

### Ejemplo 3: estilos reutilizables

```python
import tkinter as tk
from tkinter import ttk

ventana = tk.Tk()
ventana.title("Estilos del formulario")
estilo = ttk.Style(ventana)
estilo.configure("Titulo.TLabel", font=("Arial", 18, "bold"))
estilo.configure("Ayuda.TLabel", font=("Arial", 10))
estilo.configure("Principal.TButton", font=("Arial", 11, "bold"), padding=(12, 8))

panel = ttk.Frame(ventana, padding=20)
panel.grid(row=0, column=0)
ttk.Label(panel, text="Pedidos de farmacia", style="Titulo.TLabel").grid(row=0, column=0, sticky="w")
ttk.Label(panel, text="Completa los campos antes de calcular.", style="Ayuda.TLabel").grid(
    row=1, column=0, sticky="w", pady=10)
ttk.Button(panel, text="Calcular pedido", style="Principal.TButton").grid(row=2, column=0, sticky="w")
ventana.mainloop()
```

**Lectura:** el objeto `estilo` se crea después de la ventana; `.configure()` define la apariencia; el argumento `style` la asigna al control. El botón de este ejemplo solo muestra el estilo: todavía no tiene una función conectada.

## 5. Validación y proyecto guiado 
### 5.1. Del texto al dato
`Entry` y `StringVar.get()` entregan texto. Para calcular necesitamos convertirlo, pero `int("dos")` produce `ValueError`.

La validación tiene tres pasos: **leer → comprobar → utilizar**. Si hay un error, muestra una indicación concreta y detén el cálculo con `return`.

| Campo | Regla de esta actividad |
|---|---|
| Cliente | No vacío después de `strip()` |
| Teléfono | Entre 6 y 15 dígitos ASCII; sin espacios ni signos |
| Producto | Elegido del catálogo |
| Precio | Entero positivo en guaraníes, sin puntos ni comas |
| Cantidad | Entero positivo |

Estas reglas del teléfono son una simplificación del ejercicio, no una comprobación de que un número exista. Usaremos precios enteros para practicar montos en guaraníes.

### Ejemplo 4: validar una cantidad sin interfaz

```python
def validar_cantidad(texto):
    try:
        cantidad = int(texto.strip())
    except ValueError:
        raise ValueError("La cantidad debe ser un número entero.")
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    return cantidad

for dato in ("3", " 5 ", "dos", "0", "-2", "2.5"):
    try:
        print(repr(dato), "→", validar_cantidad(dato))
    except ValueError as error:
        print(repr(dato), "→ Error:", error)
```

### Lectura paso a paso
`try` intenta la conversión. `except ValueError` responde si el texto no representa un entero. Una conversión correcta no garantiza un valor permitido: por eso comprobamos también `cantidad <= 0`. `raise` comunica el error al código que llama a la función.

### 5.2. Proyecto guiado: ficha de pedido
**Construcción:** primero localiza las variables; después los contenedores; luego las funciones; finalmente los botones y el evento Enter. Ejecuta el programa completo y realiza las pruebas indicadas al final.

La función `leer_pedido()` reúne las reglas. `calcular()` muestra el resultado solo cuando todos los datos son válidos. Los importes se escriben sin separadores al ingresarlos y se presentan con puntos al mostrarlos.

```python
import tkinter as tk
from tkinter import ttk, messagebox

PRODUCTOS = ("Alcohol 70%", "Gasas", "Guantes", "Termómetro")

def moneda(valor):
    return "G. " + f"{valor:,}".replace(",", ".")

ventana = tk.Tk()
ventana.title("Farmacia – Ficha de pedido")
ventana.geometry("700x620")
ventana.minsize(580, 560)
ventana.columnconfigure(0, weight=1)
ventana.rowconfigure(0, weight=1)

estilo = ttk.Style(ventana)
estilo.configure("Titulo.TLabel", font=("Arial", 18, "bold"))
estilo.configure("Principal.TButton", font=("Arial", 11, "bold"), padding=(12, 8))

nombre_var = tk.StringVar(ventana)
telefono_var = tk.StringVar(ventana)
producto_var = tk.StringVar(ventana)
precio_var = tk.StringVar(ventana)
cantidad_var = tk.StringVar(ventana, value="1")
resumen_var = tk.StringVar(ventana, value="Completa los datos y pulsa Calcular pedido.")
estado_var = tk.StringVar(ventana, value="Listo para registrar un pedido.")

principal = ttk.Frame(ventana, padding=15)
principal.grid(row=0, column=0, sticky="nsew")
principal.columnconfigure(0, weight=1)
principal.rowconfigure(4, weight=1)
ttk.Label(principal, text="Nuevo pedido", style="Titulo.TLabel").grid(row=0, column=0, sticky="w", pady=(0, 12))

cliente = ttk.LabelFrame(principal, text="1. Datos del cliente", padding=10)
cliente.grid(row=1, column=0, sticky="ew", pady=5)
cliente.columnconfigure(1, weight=1)
ttk.Label(cliente, text="Nombre:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
entrada_nombre = ttk.Entry(cliente, textvariable=nombre_var)
entrada_nombre.grid(row=0, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(cliente, text="Teléfono:").grid(row=1, column=0, sticky="w", padx=5, pady=5)
entrada_telefono = ttk.Entry(cliente, textvariable=telefono_var)
entrada_telefono.grid(row=1, column=1, sticky="ew", padx=5, pady=5)

producto = ttk.LabelFrame(principal, text="2. Datos del producto", padding=10)
producto.grid(row=2, column=0, sticky="ew", pady=5)
producto.columnconfigure(1, weight=1)
ttk.Label(producto, text="Producto:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
combo = ttk.Combobox(producto, textvariable=producto_var, values=PRODUCTOS, state="readonly")
combo.grid(row=0, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Precio unitario (Gs.):").grid(row=1, column=0, sticky="w", padx=5, pady=5)
entrada_precio = ttk.Entry(producto, textvariable=precio_var)
entrada_precio.grid(row=1, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Cantidad:").grid(row=2, column=0, sticky="w", padx=5, pady=5)
entrada_cantidad = ttk.Entry(producto, textvariable=cantidad_var)
entrada_cantidad.grid(row=2, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Ejemplo de precio: 15000, sin puntos ni comas.").grid(
    row=3, column=0, columnspan=2, sticky="w", padx=5, pady=5)

def leer_pedido():
    nombre = nombre_var.get().strip()
    telefono = telefono_var.get().strip()
    if not nombre:
        raise ValueError("Completa el nombre del cliente.")
    if not (telefono.isascii() and telefono.isdigit() and 6 <= len(telefono) <= 15):
        raise ValueError("El teléfono debe contener entre 6 y 15 dígitos, sin signos ni espacios.")
    elegido = producto_var.get()
    if elegido not in PRODUCTOS:
        raise ValueError("Selecciona un producto del catálogo.")
    try:
        precio = int(precio_var.get().strip())
    except ValueError:
        raise ValueError("El precio debe ser un entero, por ejemplo 15000.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")
    try:
        cantidad = int(cantidad_var.get().strip())
    except ValueError:
        raise ValueError("La cantidad debe ser un entero, por ejemplo 3.")
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    return nombre, telefono, elegido, precio, cantidad

def calcular(event=None):
    # Quitamos el resumen anterior para que un error no deje visible un total viejo.
    resumen_var.set("Pedido pendiente de validación.")
    try:
        nombre, telefono, elegido, precio, cantidad = leer_pedido()
    except ValueError as error:
        estado_var.set("Revisa los datos del pedido.")
        messagebox.showwarning("Datos incompletos o incorrectos", str(error), parent=ventana)
        return
    total = precio * cantidad
    resumen_var.set(
        f"Cliente: {nombre}\nTeléfono: {telefono}\nProducto: {elegido}\n"
        f"Precio unitario: {moneda(precio)}\nCantidad: {cantidad}\nTotal: {moneda(total)}")
    estado_var.set("Cálculo correcto. El pedido todavía no se guarda en un archivo.")

def limpiar():
    nombre_var.set("")
    telefono_var.set("")
    producto_var.set("")
    precio_var.set("")
    cantidad_var.set("1")
    resumen_var.set("Completa los datos y pulsa Calcular pedido.")
    estado_var.set("Formulario limpio.")
    entrada_nombre.focus_set()

acciones = ttk.Frame(principal)
acciones.grid(row=3, column=0, sticky="ew", pady=10)
ttk.Button(acciones, text="Calcular pedido", command=calcular, style="Principal.TButton").grid(row=0, column=0, padx=(0, 8))
ttk.Button(acciones, text="Limpiar", command=limpiar).grid(row=0, column=1)

resumen = ttk.LabelFrame(principal, text="3. Resumen", padding=10)
resumen.grid(row=4, column=0, sticky="nsew", pady=5)
resumen.columnconfigure(0, weight=1)
resumen.rowconfigure(0, weight=1)
etiqueta_resumen = ttk.Label(resumen, textvariable=resumen_var, anchor="nw", justify="left", wraplength=500)
etiqueta_resumen.grid(row=0, column=0, sticky="nsew")
etiqueta_resumen.bind("<Configure>", lambda evento: etiqueta_resumen.configure(wraplength=max(100, evento.width)))
ttk.Label(principal, textvariable=estado_var, wraplength=530).grid(row=5, column=0, sticky="w", pady=(5, 0))

ventana.bind("<Return>", calcular)
entrada_nombre.focus_set()
ventana.mainloop()
```

### Explicación de las partes importantes
- **Variables:** cada `StringVar` guarda el texto de un campo; `textvariable` lo vincula con el widget.
- **Contenedores:** cliente, producto y resumen pertenecen al mismo contenedor principal, pero sus controles se distribuyen dentro de cada sección.
- **Validación:** `leer_pedido()` retorna cinco valores si todo es correcto; `raise ValueError` interrumpe la lectura ante el primer problema.
- **Respuesta al error:** `calcular()` captura el error y abre un aviso. No borra los datos ingresados.
- **Resultado:** `precio * cantidad` calcula el total; `resumen_var.set()` actualiza la etiqueta.
- **Eventos:** `event=None` permite llamar a `calcular` desde el botón y desde Enter.
- **Limpieza:** restablece la cantidad a 1 y `focus_set()` coloca el cursor en el nombre.
- **Ajuste del resumen:** `<Configure>` actualiza el ancho de ajuste del texto al redimensionar la etiqueta. No calcula pedidos.

### Pruebas guiadas
| Prueba | Datos | Resultado esperado |
|---|---|---|
| Pedido válido | Ana; 0981123456; Gasas; 15000; 3 | Total: G. 45.000 |
| Nombre vacío | Solo espacios en nombre | Aviso; conserva las entradas |
| Teléfono inválido | `abc` | Aviso sobre el teléfono |
| Producto sin elegir | Combo vacío | Solicita seleccionar un producto |
| Precio incorrecto | `15.000` | Solicita un entero sin separadores |
| Cantidad inválida | `dos`, `0`, `-2` o `2.5` | Aviso; no presenta total válido |
| Limpiar | Después de calcular | Campos vacíos; cantidad 1; foco en nombre |
| Enter | Pedido válido | Mismo cálculo que el botón |
| Cambiar tamaño | Ampliar la ventana | Entradas y resumen aprovechan el espacio |

```python
#EJERCICIO 6

import tkinter as tk
from tkinter import messagebox, ttk

def main():
    # Crea la ventana principal
    ventana = tk.Tk()
    ventana.title("Gestión de Farmacia - Descuento Comercial")
    ventana.geometry("420x450")
    
    ventana.columnconfigure(0, weight=1)
    ventana.rowconfigure(0, weight=1)
    
    contenedor = ttk.Frame(ventana, padding=15)
    contenedor.grid(row=0, column=0, sticky="nsew")
    contenedor.columnconfigure(1, weight=1)
    
    # Título
    lbl_titulo = ttk.Label(contenedor, text="Cálculo de Pedido con Descuento", font=("Arial", 12, "bold"))
    lbl_titulo.grid(row=0, column=0, columnspan=2, pady=(0, 15))
    
    # --- Sección de producto ---
    frame_producto = ttk.LabelFrame(contenedor, text="Datos del Producto", padding=10)
    frame_producto.grid(row=1, column=0, columnspan=2, sticky="ew", pady=5)
    frame_producto.columnconfigure(1, weight=1)
    
    # Precio unitario
    lbl_precio = ttk.Label(frame_producto, text="Precio (G.):")
    lbl_precio.grid(row=0, column=0, sticky="w", padx=5, pady=5)
    entry_precio = ttk.Entry(frame_producto)
    entry_precio.grid(row=0, column=1, sticky="ew", padx=5, pady=5)
    
    # Cantidad
    lbl_cantidad = ttk.Label(frame_producto, text="Cantidad:")
    lbl_cantidad.grid(row=1, column=0, sticky="w", padx=5, pady=5)
    entry_cantidad = ttk.Entry(frame_producto)
    entry_cantidad.grid(row=1, column=1, sticky="ew", padx=5, pady=5)
    
    # 1. Crea un BooleanVar y un Checkbutton dentro de la sección de producto
    var_descuento = tk.BooleanVar(value=False)
    chk_descuento = ttk.Checkbutton(
        frame_producto, 
        text="Aplicar descuento del 10 %", 
        variable=var_descuento
    )
    chk_descuento.grid(row=2, column=0, columnspan=2, sticky="w", padx=5, pady=5)
    
    # --- Resultados ---
    frame_resultados = ttk.LabelFrame(contenedor, text="Resumen", padding=10)
    frame_resultados.grid(row=2, column=0, columnspan=2, sticky="ew", pady=10)
    frame_resultados.columnconfigure(1, weight=1)
    
    lbl_res_subtotal = ttk.Label(frame_resultados, text="Subtotal: G. 0")
    lbl_res_subtotal.grid(row=0, column=0, sticky="w", padx=5, pady=2)
    
    lbl_res_descuento = ttk.Label(frame_resultados, text="Descuento: G. 0")
    lbl_res_descuento.grid(row=1, column=0, sticky="w", padx=5, pady=2)
    
    lbl_res_total = ttk.Label(frame_resultados, text="Total: G. 0", font=("Arial", 10, "bold"))
    lbl_res_total.grid(row=2, column=0, sticky="w", padx=5, pady=2)

    # Función para procesar el cálculo
    def calcular():
        try:
            precio = int(entry_precio.get().strip())
            cantidad = int(entry_cantidad.get().strip())
            
            if precio < 0 or not (1 <= cantidad <= 100):
                raise ValueError("Valores fuera de rango.")
        except ValueError:
            messagebox.showerror("Error", "Por favor, ingresa números enteros válidos (cantidad de 1 a 100).")
            return
        
        # 2. Calcula primero el subtotal
        subtotal = precio * cantidad
        
        # 3. Usa descuento = subtotal // 10 cuando la casilla esté marcada (redondea hacia abajo)
        if var_descuento.get():
            descuento = subtotal // 10
        else:
            descuento = 0
            
        # 4. Calcula total = subtotal - descuento y muestra subtotal, descuento y total
        total = subtotal - descuento
        
        lbl_res_subtotal.config(text=f"Subtotal: G. {subtotal:,}".replace(",", "."))
        lbl_res_descuento.config(text=f"Descuento: G. {descuento:,}".replace(",", "."))
        lbl_res_total.config(text=f"Total: G. {total:,}".replace(",", "."))

    # 5. Función limpiar() que también desmarca la casilla
    def limpiar():
        entry_precio.delete(0, tk.END)
        entry_cantidad.delete(0, tk.END)
        var_descuento.set(False)  # Desmarca la casilla
        lbl_res_subtotal.config(text="Subtotal: G. 0")
        lbl_res_descuento.config(text="Descuento: G. 0")
        lbl_res_total.config(text="Total: G. 0")
        entry_precio.focus()

    # Botones de acción
    btn_calcular = ttk.Button(contenedor, text="Calcular Pedido", command=calcular)
    btn_calcular.grid(row=3, column=0, sticky="ew", padx=5, pady=5)
    
    btn_limpiar = ttk.Button(contenedor, text="Limpiar", command=limpiar)
    btn_limpiar.grid(row=3, column=1, sticky="ew", padx=5, pady=5)
    
    ventana.mainloop()

if __name__ == "__main__":
    main()
```
