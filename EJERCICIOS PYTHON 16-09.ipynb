## Ejercicio 3 — Calculadora de suma

Crea dos campos numéricos y un botón **Sumar**. Muestra el resultado en una etiqueta. Si algún dato no es numérico, muestra un error.

**Pista:** utiliza `try` y `except ValueError`.

**Conceptos:** `Entry`, conversión de datos, validación y `messagebox`.

```python
import tkinter as tk
from tkinter import messagebox

def sumar():
    try:
        num1 = float(entrada1.get())
        num2 = float(entrada2.get())

        resultado = num1+num2

        etiqueta_resultado.config(text=f"Resultado: {resultado}")

    except ValueError:
        messagebox.showerror("Error", "Por favor, ingresa solo datos numericos validos.")

ventana = tk.Tk()
ventana.title("CALCULADORA DE SUMA")
ventana.geometry("300x250")
ventana.config(padx=20, pady=20)

tk.Label(ventana, text="Numero 1:").pack(anchor="w")
entrada1 = tk.Entry(ventana)
entrada1.pack(fill="x", pady=5)

tk.Label(ventana, text="Numero 2:").pack(anchor="w")
entrada2 = tk.Entry(ventana)
entrada2.pack(fill="x", pady=5)

boton_sumar = tk.Button(ventana, text="Sumar", command=sumar, bg="#4CAF50", fg="white")
boton_sumar.pack(fill="x", pady=15)

etiqueta_resultado = tk.Label(ventana, text="Resultado: ", font=("Arial", 11, "bold"))
etiqueta_resultado.pack(anchor="w", pady=5)

ventana.mainloop()
```

## Ejercicio 4 — Conversor de guaraníes a dólares

Solicita un monto en guaraníes y una cotización del dólar. Al presionar **Convertir**, calcula y muestra el equivalente en dólares con dos decimales.

Valida que:

- Ningún campo esté vacío.
- Los valores sean numéricos y mayores que cero.
- La cotización no sea cero.

**Conceptos:** formulario con `grid()`, cálculos y validación.

```python
import tkinter as tk
from tkinter import messagebox

def convertir():
    try:
        monto_str = entrada_monto.get().strip()
        cotizacion_str = entrada_cotizacion.get().strip()
        
        if not monto_str or not cotizacion_str:
            messagebox.showerror("Error", "Ningún campo puede estar vacío.")
            return
            
        monto_gs = float(monto_str)
        cotizacion = float(cotizacion_str)
        
        if monto_gs <= 0 or cotizacion <= 0:
            messagebox.showerror("Error", "Los valores deben ser numéricos y mayores que cero.")
            return
            
        resultado_usd = monto_gs / cotizacion
        
        etiqueta_resultado.config(text=f"Resultado: $ {resultado_usd:,.2f} USD")
        
    except ValueError:
        messagebox.showerror("Error", "Por favor, ingresa solo valores numéricos válidos.")

ventana = tk.Tk()
ventana.title("Ejercicio 4 — Conversor de guaraníes a dólares")
ventana.geometry("350x260")
ventana.config(padx=15, pady=15)


tk.Label(ventana, text="Monto en Guaraníes (Gs.):").grid(row=0, column=0, sticky="w", pady=5)
entrada_monto = tk.Entry(ventana, width=20)
entrada_monto.grid(row=1, column=0, sticky="we", pady=5)

tk.Label(ventana, text="Cotización del Dólar:").grid(row=2, column=0, sticky="w", pady=5)
entrada_cotizacion = tk.Entry(ventana, width=20)
entrada_cotizacion.grid(row=3, column=0, sticky="we", pady=5)

boton_convertir = tk.Button(ventana, text="Convertir", command=convertir, bg="#2196F3", fg="white")
boton_convertir.grid(row=4, column=0, sticky="we", pady=15)

etiqueta_resultado = tk.Label(ventana, text="Resultado: ", font=("Arial", 11, "bold"))
etiqueta_resultado.grid(row=5, column=0, sticky="w", pady=5)

ventana.mainloop()
```

## Ejercicio 6 — Editor de nota breve

Crea una ventana con un widget `Text` y tres botones:

- **Mostrar:** presenta el contenido en un `messagebox`.
- **Limpiar:** elimina todo el texto.
- **Salir:** solicita confirmación antes de cerrar.

**Conceptos:** `Text`, cuadros de diálogo y eventos.

```python
import tkinter as tk
from tkinter import messagebox

def mostrar_nota():
    contenido = cuadro_texto.get("1.0", "end-1c")
    
    if not contenido.strip():
        messagebox.showwarning("Advertencia", "El editor está vacío.")
    else:
        messagebox.showinfo("Contenido de la nota", contenido)

def limpiar_nota():
    cuadro_texto.delete("1.0", "end")

def salir_app():
    respuesta = messagebox.askyesno("Confirmar salida", "¿Estás seguro de que deseas salir?")
    if respuesta:
        ventana.destroy()

ventana = tk.Tk()
ventana.title("Ejercicio 6 — Editor de nota breve")
ventana.geometry("400x350")
ventana.config(padx=15, pady=15)

tk.Label(ventana, text="Escribe tu nota breve:").pack(anchor="w", pady=(0, 5))

cuadro_texto = tk.Text(ventana, width=45, height=10, font=("Arial", 10))
cuadro_texto.pack(pady=(0, 10))

frame_botones = tk.Frame(ventana)
frame_botones.pack(fill="x", pady=5)

btn_mostrar = tk.Button(frame_botones, text="Mostrar", command=mostrar_nota, bg="#4CAF50", fg="white", width=10)
btn_mostrar.pack(side="left", padx=5)

btn_limpiar = tk.Button(frame_botones, text="Limpiar", command=limpiar_nota, bg="#FF9800", fg="white", width=10)
btn_limpiar.pack(side="left", padx=5)

btn_salir = tk.Button(frame_botones, text="Salir", command=salir_app, bg="#F44336", fg="white", width=10)
btn_salir.pack(side="right", padx=5)

ventana.mainloop()
```

## Ejercicio 7 — Inicio de sesión simulado

Crea campos para usuario y contraseña. La contraseña debe mostrarse con asteriscos usando `show="*"`.

Reglas:

- Usuario correcto: `admin`.
- Contraseña correcta: `1234`.
- Si ambos coinciden, mostrar acceso concedido.
- En caso contrario, mostrar acceso denegado.
- Permitir iniciar sesión presionando Enter.

**Conceptos:** `Entry`, condicionales, `messagebox` y `bind()`.

```python
import tkinter as tk
from tkinter import messagebox

def iniciar_sesion(event=None):
    usuario = entrada_usuario.get().strip()
    contrasena = entrada_contrasena.get().strip()
    
    if not usuario or not contrasena:
        messagebox.showerror("Error", "Por favor, completa todos los campos.")
        return
        
    if usuario == "admin" and contrasena == "1234":
        messagebox.showinfo("Éxito", "Acceso concedido. ¡Bienvenido, admin!")
    else:
        messagebox.showerror("Acceso Denegado", "Usuario o contraseña incorrectos.")

ventana = tk.Tk()
ventana.title("Ejercicio 7 — Inicio de sesión simulado")
ventana.geometry("320x280")
ventana.config(padx=20, pady=20)

tk.Label(ventana, text="Usuario:").pack(anchor="w", pady=(0, 2))
entrada_usuario = tk.Entry(ventana, width=25, font=("Arial", 11))
entrada_usuario.pack(anchor="w", pady=(0, 10))

tk.Label(ventana, text="Contraseña:").pack(anchor="w", pady=(0, 2))
entrada_contrasena = tk.Entry(ventana, width=25, font=("Arial", 11), show="*")
entrada_contrasena.pack(anchor="w", pady=(0, 15))

boton_login = tk.Button(ventana, text="Iniciar sesión", command=iniciar_sesion, bg="#2196F3", fg="white", width=15)
boton_login.pack(anchor="w", pady=(0, 10))

entrada_usuario.bind("<Return>", iniciar_sesion)
entrada_contrasena.bind("<Return>", iniciar_sesion)

entrada_usuario.focus()

ventana.mainloop()
```

## Ejercicio 8 — Proyecto integrador: registro de visitantes

Construye la aplicación propuesta al inicio de la clase.

### Requisitos mínimos

1. Título y tamaño apropiado.
2. Encabezado con el nombre del sistema.
3. Campos: nombre, documento y motivo de visita.
4. Botones: Registrar, Limpiar y Salir.
5. Validación de campos obligatorios.
6. Confirmación mediante `messagebox`.
7. Limpieza automática después de registrar.
8. Cursor colocado nuevamente en el campo nombre.
9. Tecla Enter asociada a la acción Registrar.
10. Organización mediante `Frame` y `grid()`.

### Desafío adicional

Guarda temporalmente cada registro en una lista y muestra en una etiqueta cuántos visitantes fueron registrados.

```python
import tkinter as tk
from tkinter import messagebox

class RegistroVisitantesApp:
    def __init__(self, ventana):
        self.ventana = ventana
        self.ventana.title("Sistema de Control - Registro de Visitantes")
        self.ventana.geometry("450x520")
        self.ventana.config(padx=20, pady=20)
        
        self.lista_visitantes = []

        titulo_label = tk.Label(
            self.ventana, 
            text="SISTEMA DE REGISTRO DE VISITANTES", 
            font=("Arial", 14, "bold"),
            fg="#1976D2"
        )
        titulo_label.pack(pady=(0, 15))

        frame_formulario = tk.Frame(self.ventana, bd=2, relief="groove", padx=15, pady=15)
        frame_formulario.pack(fill="x", pady=5)

        tk.Label(frame_formulario, text="Nombre completo:", font=("Arial", 10, "bold")).grid(row=0, column=0, sticky="w", pady=5)
        self.entrada_nombre = tk.Entry(frame_formulario, width=30, font=("Arial", 10))
        self.entrada_nombre.grid(row=1, column=0, sticky="we", pady=(0, 10))

        tk.Label(frame_formulario, text="N° de Documento:", font=("Arial", 10, "bold")).grid(row=2, column=0, sticky="w", pady=5)
        self.entrada_documento = tk.Entry(frame_formulario, width=30, font=("Arial", 10))
        self.entrada_documento.grid(row=3, column=0, sticky="we", pady=(0, 10))

        tk.Label(frame_formulario, text="Motivo de la visita:", font=("Arial", 10, "bold")).grid(row=4, column=0, sticky="w", pady=5)
        self.entrada_motivo = tk.Entry(frame_formulario, width=30, font=("Arial", 10))
        self.entrada_motivo.grid(row=5, column=0, sticky="we", pady=(0, 10))

        frame_botones = tk.Frame(self.ventana)
        frame_botones.pack(fill="x", pady=15)

        self.btn_registrar = tk.Button(frame_botones, text="Registrar", command=self.registrar_visitante, bg="#4CAF50", fg="white", font=("Arial", 9, "bold"), width=10)
        self.btn_registrar.pack(side="left", padx=5)

        self.btn_limpiar = tk.Button(frame_botones, text="Limpiar", command=self.limpiar_campos, bg="#FF9800", fg="white", font=("Arial", 9, "bold"), width=10)
        self.btn_limpiar.pack(side="left", padx=5)

        self.btn_salir = tk.Button(frame_botones, text="Salir", command=self.salir_app, bg="#F44336", fg="white", font=("Arial", 9, "bold"), width=10)
        self.btn_salir.pack(side="right", padx=5)

        self.etiqueta_contador = tk.Label(
            self.ventana, 
            text="Total de visitantes registrados: 0", 
            font=("Arial", 10, "italic"),
            fg="#555555"
        )
        self.etiqueta_contador.pack(pady=10)

        self.entrada_nombre.bind("<Return>", self.registrar_visitante)
        self.entrada_documento.bind("<Return>", self.registrar_visitante)
        self.entrada_motivo.bind("<Return>", self.registrar_visitante)

        self.entrada_nombre.focus()

    def registrar_visitante(self, event=None):
        nombre = self.entrada_nombre.get().strip()
        documento = self.entrada_documento.get().strip()
        motivo = self.entrada_motivo.get().strip()

        if not nombre or not documento or not motivo:
            messagebox.showerror("Error de validación", "Todos los campos son obligatorios. Por favor, complétalos.")
            return

        visitante = {
            "nombre": nombre,
            "documento": documento,
            "motivo": motivo
        }
        self.lista_visitantes.append(visitante)

        messagebox.showinfo("Éxito", f"¡Visitante registrado correctamente!\n\n• Nombre: {nombre}\n• Documento: {documento}")

        total = len(self.lista_visitantes)
        self.etiqueta_contador.config(text=f"Total de visitantes registrados: {total}")
        self.limpiar_campos()

    def limpiar_campos(self):
        self.entrada_nombre.delete(0, tk.END)
        self.entrada_documento.delete(0, tk.END)
        self.entrada_motivo.delete(0, tk.END)
        self.entrada_nombre.focus()

    def salir_app(self):
        respuesta = messagebox.askyesno("Confirmar salida", "¿Estás seguro de que deseas salir del sistema?")
        if respuesta:
            self.ventana.destroy()

if __name__ == "__main__":
    root = tk.Tk()
    app = RegistroVisitantesApp(root)
    root.mainloop()
```
